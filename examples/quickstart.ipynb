```python
cities = {
    "Berlin": 3_850_000,
    "Hamburg": 1_910_000,
    "Munich": 1_590_000,
}
print(f"Loaded {len(cities)} cities")
```

## Persistent state

This cell uses `cities`, which was created in the previous cell.

```python
import datetime
largest_city = max(cities, key=cities.get)
largest_city, cities[largest_city]
```

```python
def population_table(values):
    header = f"{'City':<12} | {'Population':>12}"
    divider = "-" * len(header)
    rows = [f"{city:<12} | {population:>12,}" for city, population in values.items()]
    return "\n".join([header, divider, *rows])

print(population_table(cities))
```
