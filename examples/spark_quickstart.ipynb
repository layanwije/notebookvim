# Spark quickstart

A minimal local Spark example for `nbcli`. Run each code cell with **r**.

```python
from pyspark.sql import SparkSession 
from pyspark.sql import functions as F

spark = (
    SparkSession.builder
    .master("local[*]")
    .appName("nbcli-spak-example")
    .getOrCreate()
)

spark.sparkContext.setLogLevel("ERROR")

print(f"Spak {spark.version} is ready")
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = (
    SparkSession.builder
    .master("local[*]")
    .appName("nbcli-spark-example")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("ERROR")
print(f"Spark {spark.version} is ready")
```

```python
sales = [
    ("Sweden", 12831),
    ("Germany", 10932),
    ("Norway", 7821),
    ("Sweden", 4100),
]

df = spark.createDataFrame(sales, ["country", "sales"])
df.show(truncate=False)
```

```python
totals = (
    df.groupBy("country")
      .agg(F.sum("sales").alias("total_sales"))
      .orderBy(F.desc("total_sales"))
)
totals.show(truncate=False)
```
