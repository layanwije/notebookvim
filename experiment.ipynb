# Spark quick Start

This is to go over examples of spark


```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = (
    SparkSession.builder
    .master("local[*]")
    .appName("nbcli-spark-example")
    .getOrCreate()
)

spark.sparkContext.setLogLevel("ERROR")
print(f"Spark {spark.version} is ready")
```

```python
sales = [
    ("Sweden", 12831),
    ("Germany", 10932),
    ("Norway", 7821),
    ("Sweden", 4100),
]

df = spark.createDataFrame(sales, ["country", "sales"])
df.show(truncate=False)
```
